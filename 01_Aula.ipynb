```python
import re
import nltk
import spacy

nltk.download("punkt")
nltk.download("rslp")
nltk.download("punkt_tab")

nlp = spacy.load("pt_core_news_sm")
```

```python
texto_bruto = """
<p>O dev falou que a API gen_ai_v2 tá quebrando no deploy!!! 🤬</p>
Verificar o bug na linha 404... acessar http://github.com/projeto/issue/99
A documentação diz que os desenvolvedores estudaram o código, mas a máquina não compreendeu as regras simbólicas.
"""
```

```python
print("TEXTO BRUTO")
print(texto_bruto)
```

```python
print("REGEX 01: remove tgs HTML")
texto_limpo = re.sub(r'<.*?>', '', texto_bruto)
print(texto_limpo)

print("------\n")

print("REGEX 02: remove URLs")
texto_limpo = re.sub(r'http\S+', '', texto_limpo)
print(texto_limpo)

print("------\n")

print("REGEX 03: manter apenas letras e espaços, removendo números e pontuaçoes")
texto_limpo = re.sub(r'[^a-zA-ZáéíóúãõâêîôûçÁÉÍÓÚÃÕÂÊÎÔÛÇ\s]', '', texto_limpo)
print(texto_limpo)

print("------\n")

print("REGEX 04: remover espaçamento desnecessario\n")
texto_limpo = re.sub(r'\s+', ' ', texto_limpo).strip()
print(texto_limpo)
```

```python
from nltk.tokenize import word_tokenize

# realiza a tokenização
tokens = word_tokenize(texto_limpo, language='portuguese')
print(f"Total de tokens necessários encontrados: {len(tokens)}\n")
print(tokens)

# aqui imprime cada token em uma linha
#for token in tokens:
    #print(token)
```

```python
stemmer = nltk.stem.RSLPStemmer()

print("COMPARATIVO: Palavra Original/ Stemming/ Lemmetization\n")

# Stem: reduz as palavras ao radical, sem considerar contexto. Perde estrutura e sentido natural.
# Lema: transforma cada palavra na forma base correta. Ela perde concordância natural, mas mantém o significado base correto.

doc_spacy = nlp(texto_limpo)

for token in doc_spacy:
  if token.text.strip():
    palavra_original = token.text
    palavra_stem = stemmer.stem(palavra_original)
    palavra_lemma = token.lemma_

    print(f"Original: {palavra_original:<15} | Stem: {palavra_stem:<12} | Lema: {palavra_lemma}")
```

```python
print("CLASSIFICAÇÃO GRAMATICAL (POS Tagging)\n")

# POS Tagging: indica a classe gramatical da palavra, ou seja, o que a palavra é na frase

for token in doc_spacy:
  if token.text.strip():
    print(f"Palavra: {token.text:<15} -> Tag: {token.pos_}")
```

```python
from collections import Counter

traducao_tags = {
    'NOUN': 'Substantivos',
    'VERB': 'Verbos',
    'ADJ': 'Adjetivos',
    'ADV': 'Advérbios',
    'PRON': 'Pronomes',
    'DET': 'Artigos/Determinantes',
    'ADP': 'Preposições',
    'PROPN': 'Nomes Próprios',
    'AUX': 'Verbos Auxiliares',
    'CCONJ': 'Conjunções Coordenativas',
    'SCONJ': 'Conjunções Subordinativas',
    'PUNCT': 'Pontuação'
}

tags_encontradas = [token.pos_ for token in doc_spacy if token.text.strip()]
contagem = Counter(tags_encontradas)

print("ESTATÍSTICAS DO TEXTO\n")

for tag, quantidade in contagem.most_common():
    nome_pt = traducao_tags.get(tag, tag)  # se não existir no dicionário, mantém a tag original
    print(f"{nome_pt:<25} -> {quantidade} ocorrência(s)")
```
